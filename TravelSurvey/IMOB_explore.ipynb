# IMOB2017 exploration and visualizations

This notebook loads the preprocessed IMOB outputs (parquet/jsonl) produced by `TravelSurvey/load_imob.py` and provides quick visualizations for persona and trip features useful for LLM/persona work.

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
sns.set(style='whitegrid')
```

```python
# locate data produced by TravelSurvey/load_imob.py (default folder: TravelSurvey/data)
base = Path("..").resolve().parent
data_dir = base / 'data'
# fallback to repository Population raw location used earlier
if not data_dir.exists():
    data_dir = base.parent / 'Population' / '.data' / 'IMOB2017'
persons_path = data_dir / 'imob_persons.parquet'
trips_path = data_dir / 'imob_trips.parquet'
# if not present, try train split folder
if not persons_path.exists():
    persons_path = data_dir / 'train' / 'imob_persons.parquet'
if not trips_path.exists():
    trips_path = data_dir / 'train' / 'imob_trips.parquet'
print('Persons file:', persons_path)
print('Trips file:  ', trips_path)
persons = pd.read_parquet(persons_path) if persons_path.exists() else pd.DataFrame()
trips = pd.read_parquet(trips_path) if trips_path.exists() else pd.DataFrame()
print('Loaded shapes -> persons:', persons.shape, 'trips:', trips.shape)
```

```python
# Preview and simple visualizations
if not persons.empty:
    display(persons.head())
    plt.figure(figsize=(8,4))
    persons['gender'].value_counts().plot(kind='bar', title='Gender distribution')
    plt.tight_layout(); plt.show()
    plt.figure(figsize=(10,4))
    persons['ageGroup'].value_counts().sort_index().plot(kind='bar', title='Age group distribution')
    plt.tight_layout(); plt.show()
else:
    print('No persons data available')
if not trips.empty:
    display(trips.head())
    plt.figure(figsize=(8,4))
    trips['mode'].value_counts().plot(kind='bar', title='Mode share (legs)')
    plt.tight_layout(); plt.show()
    plt.figure(figsize=(8,4))
    trips['distance'].dropna().astype(float).div(1000).hist(bins=50)
    plt.xlabel('Distance (km)')
    plt.title('Trip distance distribution')
    plt.tight_layout(); plt.show()
    plt.figure(figsize=(8,4))
    trips.groupby('person_id').size().hist(bins=50)
    plt.xlabel('Trips per person')
    plt.title('Trips per person distribution')
    plt.tight_layout(); plt.show()
else:
    print('No trips data available')
```
